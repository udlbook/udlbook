```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from numpy import asarray
```

```python
# Define 4 by 4 original patch
orig_4_4 = np.array([[1, 3, 5,3 ], [6,2,0,8], [4,6,1,4], [2,8,0,3]])
print(orig_4_4)
```

```python
def downsample(x_in):
  x_out = np.zeros(( int(np.ceil(x_in.shape[0]/2)), int(np.ceil(x_in.shape[1]/2)) ))
  # TODO -- write the downsampling routine
  # Take every other pixel in both dimensions
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = x_in[i*2, j*2]

  return x_out
```

```python
print("Original:")
print(orig_4_4)
print("Downsampled:")
print(downsample(orig_4_4))
```

```python
# Now let's try max-pooling
def maxpool(x_in):
  x_out = np.zeros(( int(np.floor(x_in.shape[0]/2)), int(np.floor(x_in.shape[1]/2)) ))
  # TODO -- write the maxpool routine
  # Take maximum of each 2x2 block
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = np.max(x_in[i*2:i*2+2, j*2:j*2+2])

  return x_out
```

```python
# Finally, let's try mean pooling
def meanpool(x_in):
  x_out = np.zeros(( int(np.floor(x_in.shape[0]/2)), int(np.floor(x_in.shape[1]/2)) ))
  # TODO -- write the meanpool routine
  # Take mean of each 2x2 block
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = np.mean(x_in[i*2:i*2+2, j*2:j*2+2])

  return x_out
```

```python
# Define 2 by 2 original patch
orig_2_2 = np.array([[6, 8], [8,4]])
print(orig_2_2)
```

```python
# Let's first use the duplication method
def duplicate(x_in):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  # TODO -- write the duplication routine
  # Duplicate each pixel into a 2x2 block
  for i in range(x_in.shape[0]):
    for j in range(x_in.shape[1]):
      x_out[i*2:i*2+2, j*2:j*2+2] = x_in[i, j]

  return x_out
```

```python
print("Original:")
print(orig_2_2)
print("Duplicated:")
print(duplicate(orig_2_2))
```

```python
# Now let's try max pooling back up
# The input x_high_res is the original high res image, from which you can deduce the position of the maximum index
def max_unpool(x_in, x_high_res):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  # TODO -- write the unpooling routine
  # Place each value at the position of the max in the corresponding 2x2 block from high_res
  for i in range(x_in.shape[0]):
    for j in range(x_in.shape[1]):
      # Find the position of max in the 2x2 block of high_res
      block = x_high_res[i*2:i*2+2, j*2:j*2+2]
      max_pos = np.unravel_index(np.argmax(block), block.shape)
      x_out[i*2 + max_pos[0], j*2 + max_pos[1]] = x_in[i, j]

  return x_out
```

Finally, we'll try upsampling using bilinear interpolation.  We'll treat the positions off the image as zeros by padding the original image and round fractional values upwards using np.ceil()

```python
def bilinear(x_in):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  x_in_pad = np.zeros((x_in.shape[0]+1, x_in.shape[1]+1))
  x_in_pad[0:x_in.shape[0],0:x_in.shape[1]] = x_in
  # TODO -- write the duplication routine
  # Bilinear interpolation: compute weighted average of 4 surrounding pixels
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      # Position in the padded input (scaled down by 2)
      in_i = i / 2.0
      in_j = j / 2.0
      # Get floor and ceil indices
      i_floor = int(np.floor(in_i))
      j_floor = int(np.floor(in_j))
      i_ceil = int(np.ceil(in_i))
      j_ceil = int(np.ceil(in_j))
      # Weights for bilinear interpolation
      w_i = in_i - i_floor
      w_j = in_j - j_floor
      # Bilinear interpolation
      x_out[i, j] = (1 - w_i) * (1 - w_j) * x_in_pad[i_floor, j_floor] + \
                    w_i * (1 - w_j) * x_in_pad[i_ceil, j_floor] + \
                    (1 - w_i) * w_j * x_in_pad[i_floor, j_ceil] + \
                    w_i * w_j * x_in_pad[i_ceil, j_ceil]

  return x_out
```
