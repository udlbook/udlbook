```python
# Define b to be the verticalconcatenation of p and q
b = np.hstack((p,q))[np.newaxis].transpose()
```

```python
# TODO:  Now construct the matrix A that has the initial distribution constraints
# so that A @ TPFlat=b where TPFlat is the transport plan TP vectorized rows first so TPFlat = np.flatten(TP)
# The constraints are:
# - For each i (0 to 9): sum_j TP[i,j] = p[i]  (row sums)
# - For each j (0 to 9): sum_i TP[i,j] = q[j]  (column sums)
A = np.zeros((20,100))
# Row constraints: sum over columns for each row
for i in range(10):
  for j in range(10):
    A[i, i*10 + j] = 1
# Column constraints: sum over rows for each column
for j in range(10):
  for i in range(10):
    A[10+j, i*10 + j] = 1
```

Now we have all of the things we need.  The vectorized distance matrix $\mathbf{c}$,  the constraint matrix $\mathbf{A}$, the vectorized and concatenated original distribution $\mathbf{b}$.  We can run the linear programming optimization.

```python
# We don't need the constraint that p>0 as this is the default
opt = linprog(c, A_eq=A, b_eq=b)
print(opt)
```

Extract the answer and display

```python
TP = np.array(opt.x).reshape(10,10)
draw_2D_heatmap(TP,r'Transport plan $\mathbf{P}$', my_colormap)
```

Compute the Wasserstein distance


```python
was = np.sum(TP * dist_mat)
print("Your Wasserstein distance = ", was)
print("Correct answer =  0.15148578811369506")
```

*   Reverse KL divergence $D_{KL}[q,p]$ between these distributions
*  Jensen-Shannon divergence $D_{JS}[p,q]$ between these distributions

What do you conclude?

```python
# Compute KL divergences
eps = 1e-10

# Forward KL: D_KL[p||q]
forward_kl = np.sum(p * np.log((p + eps) / (q + eps)))

# Reverse KL: D_KL[q||p]
reverse_kl = np.sum(q * np.log((q + eps) / (p + eps)))

# Jensen-Shannon: 0.5*KL[p||m] + 0.5*KL[q||m]
m = (p + q) / 2
js_div = 0.5 * np.sum(p * np.log((p + eps) / (m + eps))) + 0.5 * np.sum(q * np.log((q + eps) / (m + eps)))

print("\nDivergence Comparisons:")
print(f"Wasserstein distance = {was:.10f}")
print(f"Forward KL [p||q]    = {forward_kl:.10f}")
print(f"Reverse KL [q||p]    = {reverse_kl:.10f}")
print(f"Jensen-Shannon [p,q] = {js_div:.10f}")
print("\nConclusion:")
print("The Wasserstein distance is more suitable for GAN training because:")
print("1. It has better gradient properties (never zero when gradients exist)")
print("2. It provides a geometric interpretation (cost of transport)")
print("3. It's differentiable and enables stable training")
print("4. KL divergences are 0 only if distributions are identical,")
print("   but Wasserstein provides smooth gradients even when supports don't overlap")
```
